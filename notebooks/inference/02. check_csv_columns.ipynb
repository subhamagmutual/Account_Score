# Diagnostic: What Columns Are In Your CSV?

```python
import pandas as pd

csv_path = r'C:\Box\Box\BOX Subhashree Singh\Business\PAS\physician_scores_20260429_141959.csv'

print("Loading CSV...")
df = pd.read_csv(csv_path, low_memory=False)

print(f"\n✅ Loaded {len(df):,} rows, {len(df.columns)} columns\n")

print("ALL COLUMNS IN YOUR CSV:")
print("="*80)
for i, col in enumerate(df.columns, 1):
    print(f"{i:3d}. {col}")
```

```python
# Look for composite score column
print("\n\nSearching for composite score column...")
print("="*80)

score_cols = [col for col in df.columns if 'score' in col.lower() or 'composite' in col.lower()]

if score_cols:
    print(f"Found columns with 'score' or 'composite':")
    for col in score_cols:
        print(f"  - {col}")
else:
    print("No columns with 'score' or 'composite' in the name found.")
    print("\nSearching for other likely candidates...")
    for col in df.columns[:20]:
        print(f"  - {col}")
```
