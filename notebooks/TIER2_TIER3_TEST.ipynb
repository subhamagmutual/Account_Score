# TIER 2: Folder Structure, CI/CD, Documentation

## Test 1: Folder Structure Verification

````python
# Test 1: Check all required folders exist
required_folders = [
    'src/account_score',
    'tests',
    'config',
    'docs',
    'docs/api',
    'docs/deployment',
    'docs/data-governance',
    'docs/history',
    'scripts',
    'models',
    'logs',
    '.github/workflows'
]

print("\n📁 FOLDER STRUCTURE CHECK")
print("=" * 50)

all_folders_exist = True
for folder in required_folders:
    path = PROJECT_ROOT / folder
    exists = path.exists() and path.is_dir()
    status = "✓" if exists else "✗"
    print(f"{status} {folder}")
    if not exists:
        all_folders_exist = False

if all_folders_exist:
    print("\n✓ All required folders present!")
else:
    print("\n✗ Some folders are missing")
````

## Test 2: Source Code Organization

````python
# Test 2: Check src/account_score/ modules
print("\n📦 SOURCE CODE MODULES")
print("=" * 50)

src_dir = PROJECT_ROOT / 'src' / 'account_score'
if src_dir.exists():
    py_files = sorted([f.name for f in src_dir.glob('*.py')])
    print(f"Found {len(py_files)} Python modules:\n")
    for f in py_files:
        print(f"  • {f}")
    
    expected_modules = ['__init__.py', 'pipeline.py', 'binner.py', 'scorer.py']
    missing = [m for m in expected_modules if m not in py_files]
    if missing:
        print(f"\n✗ Missing modules: {missing}")
    else:
        print(f"\n✓ All core modules present")
else:
    print(f"✗ src/account_score/ not found")
````

## Test 3: Documentation Completeness

````python
# Test 3: Check documentation files
print("\n📚 DOCUMENTATION FILES")
print("=" * 50)

doc_files = {
    'API Reference': 'docs/api/API_REFERENCE.md',
    'Deployment Checklist': 'docs/deployment/DEPLOYMENT_CHECKLIST.md',
    'Data Dictionary': 'docs/data-governance/DATA_DICTIONARY.md',
    'Data Lineage': 'docs/data-governance/DATA_LINEAGE.md',
    'TIER 1 Summary': 'docs/history/TIER1_IMPLEMENTATION_SUMMARY.md',
}

doc_status = {}
for name, path in doc_files.items():
    full_path = PROJECT_ROOT / path
    exists = full_path.exists()
    size = full_path.stat().st_size if exists else 0
    status = "✓" if exists and size > 1000 else "✗"
    print(f"{status} {name:25} ({size:,} bytes)")
    doc_status[name] = exists and size > 1000

if all(doc_status.values()):
    print("\n✓ All documentation files complete")
else:
    print("\n✗ Some documentation files are missing or incomplete")
````

## Test 4: Configuration Files

````python
# Test 4: Check configuration files
print("\n⚙️  CONFIGURATION FILES")
print("=" * 50)

config_files = [
    'config/binning_thresholds.json',
    'config/pipeline_params.json',
    'config/scoring_weights.json',
    'config/specialty_tiers.json',
    'config/state_tiers.json',
    'pyproject.toml',
    'pytest.ini',
    '.gitignore',
]

config_status = {}
for config in config_files:
    path = PROJECT_ROOT / config
    exists = path.exists()
    status = "✓" if exists else "✗"
    print(f"{status} {config}")
    config_status[config] = exists
    
    # Try to validate JSON files
    if exists and config.endswith('.json'):
        try:
            with open(path, 'r') as f:
                json.load(f)
            print(f"   → Valid JSON")
        except json.JSONDecodeError as e:
            print(f"   → ✗ Invalid JSON: {e}")

if all(config_status.values()):
    print("\n✓ All configuration files present")
else:
    print("\n✗ Some configuration files are missing")
````

## Test 5: CI/CD Workflows

````python
# Test 5: Check CI/CD workflows
print("\n🔄 CI/CD WORKFLOWS")
print("=" * 50)

workflows_dir = PROJECT_ROOT / '.github' / 'workflows'
if workflows_dir.exists():
    workflow_files = sorted(workflows_dir.glob('*.yml'))
    print(f"Found {len(workflow_files)} workflow files:\n")
    
    expected_workflows = {'tests.yml', 'lint.yml', 'build.yml'}
    actual_workflows = {f.name for f in workflow_files}
    
    for workflow in workflow_files:
        print(f"  ✓ {workflow.name}")
    
    missing = expected_workflows - actual_workflows
    if missing:
        print(f"\n✗ Missing workflows: {missing}")
    else:
        print(f"\n✓ All expected workflows present")
else:
    print("✗ .github/workflows/ directory not found")
````

## Test 6: Import Resolution

````python
# Test 6: Verify imports work with new structure
print("\n🔗 IMPORT RESOLUTION")
print("=" * 50)

# Add project root to sys.path if not already there
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import_tests = [
    'from src.account_score import pipeline',
    'from src.account_score.binner import BinFitter, BinApplier',
    'from src.account_score.scorer import ScoreCalculator',
    'from src.account_score.config_validator import ConfigValidator',
]

import_status = {}
for import_stmt in import_tests:
    try:
        exec(import_stmt)
        status = "✓"
        message = "Success"
    except Exception as e:
        status = "✗"
        message = str(e)[:50]
    
    print(f"{status} {import_stmt}")
    if status == "✗":
        print(f"   → Error: {message}")
    import_status[import_stmt] = status == "✓"

if all(import_status.values()):
    print("\n✓ All imports resolve correctly")
else:
    print("\n✗ Some imports failed")
````

## Test 7: Run Test Suite

````python
# Test 7: Run pytest
print("\n🧪 RUNNING PYTEST")
print("=" * 50)
print("This may take a minute...\n")

try:
    result = subprocess.run(
        ['pytest', 'tests/', '-v', '--tb=short', '--co', '-q'],
        cwd=PROJECT_ROOT,
        capture_output=True,
        text=True,
        timeout=30
    )
    
    print("Test discovery:")
    print(result.stdout[:500])
    
    if result.returncode == 0:
        # Count tests
        test_count = result.stdout.count('::test_')
        print(f"\n✓ Found {test_count} tests")
    else:
        print(f"\n✗ Test discovery failed")
        if result.stderr:
            print(f"Error: {result.stderr[:200]}")
            
except subprocess.TimeoutExpired:
    print("✗ Test discovery timed out")
except FileNotFoundError:
    print("✗ pytest not installed - run 'pip install pytest' first")
````

# TIER 3: Makefile & Pre-commit Hooks

## Test 8: Makefile Existence & Content

````python
# Test 8: Check Makefile
print("\n📝 MAKEFILE VERIFICATION")
print("=" * 50)

makefile_path = PROJECT_ROOT / 'Makefile'
if makefile_path.exists():
    print("✓ Makefile exists\n")
    
    with open(makefile_path, 'r') as f:
        content = f.read()
    
    # Check for targets
    expected_targets = ['install', 'install-dev', 'test', 'lint', 'format', 'clean', 'help']
    found_targets = []
    
    print("Targets found:")
    for target in expected_targets:
        if f'{target}:' in content:
            print(f"  ✓ {target}")
            found_targets.append(target)
        else:
            print(f"  ✗ {target}")
    
    if len(found_targets) == len(expected_targets):
        print(f"\n✓ All {len(expected_targets)} targets present")
    else:
        print(f"\n✗ Missing {len(expected_targets) - len(found_targets)} targets")
else:
    print("✗ Makefile not found")
````

## Test 9: Makefile Commands

````python
# Test 9: Test Makefile help command (or validate content on Windows)
print("\n🎯 MAKEFILE HELP OUTPUT")
print("=" * 50)

try:
    result = subprocess.run(
        ['make', 'help'],
        cwd=PROJECT_ROOT,
        capture_output=True,
        text=True,
        timeout=10
    )
    
    if result.returncode == 0:
        print(result.stdout)
        print("\n✓ Makefile help works")
    else:
        print(f"✗ Makefile error: {result.stderr}")
except FileNotFoundError:
    print("ℹ 'make' command not found (normal on Windows)")
    print("\nValidating Makefile content instead...\n")
    
    makefile_path = PROJECT_ROOT / 'Makefile'
    if makefile_path.exists():
        with open(makefile_path, 'r') as f:
            content = f.read()
        
        # Extract help section
        if '@echo' in content:
            lines = content.split('\n')
            in_help = False
            for line in lines:
                if 'help:' in line or '@echo' in line:
                    in_help = True
                if in_help and ('Available targets:' in line or 'make' in line):
                    print(line)
                if in_help and line.strip() == '' and '@echo' not in line:
                    break
        
        print("\n✓ Makefile content is valid")
        print("  → To use: Install GNU Make or use WSL/Git Bash on Windows")
    else:
        print("✗ Makefile not found")
except subprocess.TimeoutExpired:
    print("✗ Makefile help timed out")
except Exception as e:
    print(f"✗ Error: {e}")
````

## Test 10: Pre-commit Configuration

````python
# Test 10: Check pre-commit configuration
print("\n🔒 PRE-COMMIT HOOKS CONFIGURATION")
print("=" * 50)

precommit_config = PROJECT_ROOT / '.pre-commit-config.yaml'
if precommit_config.exists():
    print("✓ .pre-commit-config.yaml exists\n")
    
    with open(precommit_config, 'r') as f:
        content = f.read()
    
    # Check for common hooks
    expected_hooks = ['black', 'isort', 'flake8', 'mypy']
    print("Hooks configured:")
    for hook in expected_hooks:
        if hook in content:
            print(f"  ✓ {hook}")
        else:
            print(f"  ✗ {hook}")
    
    print("\n✓ Pre-commit configuration present")
else:
    print("✗ .pre-commit-config.yaml not found")
````

## Test 11: Pre-commit Installation Check

````python
# Test 11: Check if pre-commit is installed
print("\n📦 PRE-COMMIT TOOL CHECK")
print("=" * 50)

try:
    result = subprocess.run(
        ['pre-commit', '--version'],
        capture_output=True,
        text=True,
        timeout=5
    )
    
    if result.returncode == 0:
        print(f"✓ pre-commit installed: {result.stdout.strip()}")
    else:
        print("✗ pre-commit command failed")
        print("  → Install with: pip install pre-commit")
except FileNotFoundError:
    print("✗ pre-commit not installed")
    print("  → Install with: pip install pre-commit")
except Exception as e:
    print(f"✗ Error: {e}")
````

## Test 12: Pre-commit Hooks Installation

````python
# Test 12: Check if hooks are installed in .git
print("\n🔗 PRE-COMMIT HOOKS INSTALLATION")
print("=" * 50)

git_hooks_dir = PROJECT_ROOT / '.git' / 'hooks'
if git_hooks_dir.exists():
    pre_commit_hook = git_hooks_dir / 'pre-commit'
    if pre_commit_hook.exists():
        print("✓ Pre-commit hook installed in .git/hooks/")
        with open(pre_commit_hook, 'r') as f:
            content = f.read()
        if 'pre-commit' in content:
            print("✓ Hook file references pre-commit")
        else:
            print("✗ Hook file doesn't reference pre-commit")
    else:
        print("✗ Pre-commit hook not installed")
        print("  → Install with: pre-commit install")
else:
    print("✗ .git/hooks directory not found")
    print("  → This is a git repository issue")
````
