## Test 3: Batch Scoring via API

```python
print('\n' + '='*70)
print('TEST 3: BATCH SCORING VIA API')
print('='*70)

# Create batch request
batch_physicians = pd.DataFrame({
    'npi': np.random.randint(1000000000, 9999999999, 100),
    'specialty': np.random.choice(['Surgery', 'Medicine', 'Pediatrics'], 100),
    'state': np.random.choice(['CA', 'TX', 'NY'], 100),
})

batch_request = {
    'physicians': batch_physicians.to_dict(orient='records'),
    'max_records': 100,
}

# Simulate batch response
batch_response = {
    'total_records': 100,
    'successful': 100,
    'failed': 0,
    'processing_time_ms': 245,
    'scores': [
        {
            'npi': int(row['npi']),
            'composite_score': np.random.uniform(1, 10),
        }
        for _, row in batch_physicians.head(5).iterrows()
    ],
}

print(f'\nBatch Request: {len(batch_physicians)} physicians')
print(f'\nBatch Response Summary:')
print(f'  Total: {batch_response["total_records"]}')
print(f'  Successful: {batch_response["successful"]}')
print(f'  Failed: {batch_response["failed"]}')
print(f'  Processing time: {batch_response["processing_time_ms"]}ms')
print(f'  Throughput: {batch_response["total_records"] / (batch_response["processing_time_ms"] / 1000):.0f} scores/sec')
print(f'\n[OK] Batch API working, throughput: {batch_response["total_records"] / (batch_response["processing_time_ms"] / 1000):.0f} scores/sec')
```
