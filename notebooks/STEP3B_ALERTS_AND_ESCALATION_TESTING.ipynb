## Alert Management System

This notebook demonstrates:
1. Alert creation and management
2. Severity-based escalation
3. Email and Slack notifications
4. Alert acknowledgment tracking
5. Incident response procedures

```python
import sys
sys.path.insert(0, '../')

from datetime import datetime
import json

print('[OK] Step 3B: Alerts & Escalation')
print(f'Date: {datetime.now().strftime("%Y-%m-%d")}')
```

## Test 1: Define Alert Rules

```python
print('='*70)
print('TEST 1: DEFINE ALERT RULES')
print('='*70)

alert_rules = [
    {
        'name': 'PSI Drift Alert',
        'condition': 'psi > 0.25',
        'severity': 'BLOCKER',
        'action': 'Block deployment, investigate drift',
        'escalate_to': ['ops@magmutual.com', 'slack:data-science'],
    },
    {
        'name': 'Distribution Shift Alert',
        'condition': 'mean_shift > 5%',
        'severity': 'WARNING',
        'action': 'Alert team, monitor closely',
        'escalate_to': ['slack:data-science'],
    },
    {
        'name': 'Data Quality Alert',
        'condition': 'null_rate > 5%',
        'severity': 'BLOCKER',
        'action': 'Block scoring, fix data source',
        'escalate_to': ['ops@magmutual.com', 'slack:devops'],
    },
    {
        'name': 'API Latency Alert',
        'condition': 'latency > 200ms',
        'severity': 'WARNING',
        'action': 'Scale API, optimize queries',
        'escalate_to': ['slack:devops'],
    },
]

print(f'\nAlert Rules Configured:')
for i, rule in enumerate(alert_rules, 1):
    print(f'\n{i}. {rule["name"]}')
    print(f'   Condition: {rule["condition"]}')
    print(f'   Severity: {rule["severity"]}')
    print(f'   Escalate to: {', '.join(rule["escalate_to"])}')
```

## Test 2: Generate Alert

```python
print('\n' + '='*70)
print('TEST 2: GENERATE ALERT')
print('='*70)

alert = {
    'alert_id': 'ALERT-20260728-001',
    'timestamp': datetime.now().isoformat(),
    'rule': 'Distribution Shift Alert',
    'severity': 'WARNING',
    'message': 'Distribution shift detected: mean +3.2% from baseline',
    'details': {
        'baseline_mean': 5.5,
        'current_mean': 5.68,
        'shift_pct': 3.2,
    },
    'status': 'OPEN',
    'escalated_to': ['slack:data-science'],
}

print(f'\nAlert Generated:')
print(json.dumps(alert, indent=2))
```

## Test 3: Escalation Matrix

```python
print('\n' + '='*70)
print('TEST 3: ESCALATION MATRIX')
print('='*70)

escalation_matrix = {
    'INFO': {
        'channels': ['logs'],
        'delay': '0 min',
        'description': 'Log only, no alert',
    },
    'WARNING': {
        'channels': ['slack:data-science'],
        'delay': '0 min',
        'escalate_after': '1 hour',
        'description': 'Alert team, monitor',
    },
    'BLOCKER': {
        'channels': ['email:ops', 'slack:data-science', 'pagerduty'],
        'delay': '0 min',
        'escalate_after': '15 min',
        'description': 'Block deployment, escalate immediately',
    },
}

print(f'\nEscalation Matrix:')
for severity, config in escalation_matrix.items():
    print(f'\n{severity}:')
    for key, val in config.items():
        if key == 'channels':
            print(f'  {key}: {', '.join(val)}')
        else:
            print(f'  {key}: {val}')
```
